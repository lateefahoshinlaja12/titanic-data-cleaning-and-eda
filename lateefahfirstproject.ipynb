```python
#importing these libraries to help plot graphs for the data#
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Read the titanic csv file using pandas#
df = pd.read_csv("titanic.csv")
```

```python
#HANDLING MISSING DATA#
#while handling the missing data column for cabin...i experienced and error stating 'KeyError: "['Cabin'] not found in axis'so i searched it up,
#it was suggested i add a safety parameter which tells Python: Look for the Cabin column and drop it, but if it is already deleted or gone, just move on to the next line without throwing an error

df['Age'] = df['Age'].fillna(df['Age'].median())
df = df.drop(columns=['Cabin'], errors='ignore')
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
print(df.isnull().sum())
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Survived', hue='Sex', data=df, palette='Set2')
plt.title('Titanic Survival by Gender (0 = No, 1 = Yes)')
plt.xlabel('Survived')
plt.ylabel('Passenger Count')
plt.show()
```
